```python

# UCS420 Assignment 2
# Question 1 - Lists
# Name : Hazel Latka
# Roll No : 1024160105
# Branch : COSE


# Taking individual digits of roll number
roll_no = "1024160105"

# Multiplying each digit by 10
L = [int(digit) * 10 for digit in roll_no]

print("Original List L:")
print(L)
```

```python
# Adding a number using append()

L.append(99)

print("After append(99):")
print(L)

# append() adds the element at the END of the list.
```

```python
# Adding a number using insert()

L.insert(3, 55)

print("After insert(3,55):")
print(L)

# insert(position, value) adds the element at the specified position.
```

```python
# Removing an element using remove()

L.remove(0)

print("After remove(0):")
print(L)

# remove() deletes the FIRST occurrence of the specified value.
```

```python
# Removing an element using pop()

removed = L.pop()

print("Removed Element =", removed)

print("After pop():")
print(L)

# pop() removes the LAST element if no index is given.
```

```python
# Sorting in ascending order

L.sort()

print("Ascending Order:")
print(L)
```

```python
# Sorting in descending order

L.sort(reverse=True)

print("Descending Order:")
print(L)
```

```python
print("First Three Elements:")
print(L[:3])

print("Last Three Elements:")
print(L[-3:])
```

```python
# Finding average of list

average = sum(L) / len(L)

print("Average =", average)

# Creating a new list having elements greater than average

greater_than_average = [x for x in L if x > average]

print("Elements greater than average:")
print(greater_than_average)
```

Explanation
append()
Adds an element at the end of the list.
insert(index, value)
Inserts an element at a specified position.
remove(value)
Removes the first occurrence of the given value.
pop()
Removes and returns an element.
By default, it removes the last element.
sort()
Arranges the list in ascending order.
sort(reverse=True)
Arranges the list in descending order.
List Slicing
L[:3] → First three elements.
L[-3:] → Last three elements.
List Comprehension

Creates a new list using a single line of code based on a condition.

greater_than_average = [x for x in L if x > average]

This creates a new list containing only elements greater than the average.

```python

# UCS420 Assignment 2
# Question 2 - Tuples
# Name : Hazel Latka
# Roll No : 1024160105
# Branch : COSE


# Creating a tuple from the roll number digits

roll_no = "1024160105"

T = tuple(int(digit) for digit in roll_no)

print("Tuple T:")
print(T)
```

```python
# Finding the maximum and minimum elements

maximum = max(T)
minimum = min(T)

print("Maximum Element =", maximum)
print("Minimum Element =", minimum)
```

```python
# Counting frequency of each element

print("Frequency of each element:\n")

for item in sorted(set(T)):
    print(item, "appears", T.count(item), "time(s)")
```

```python
# Another method using dictionary

frequency = {}

for item in T:
    frequency[item] = frequency.get(item, 0) + 1

print("Frequency Dictionary:")
print(frequency)
```

Explanation
Tuple
A tuple is an ordered collection of elements.
Tuples are immutable, meaning their values cannot be changed after creation.

Example:

T = (1, 2, 3)
max()

Returns the largest element.

max(T)
min()

Returns the smallest element.

min(T)
count()

Counts how many times an element appears.

Example:

T.count(1)

Output:

3
set(T)

Removes duplicate values from the tuple.

Example:

set(T)

Output:

{0, 1, 2, 4, 5, 6}
Dictionary Method

Stores each unique element as a key and its frequency as the corresponding value.

frequency[item] = frequency.get(item, 0) + 1
get(item, 0) returns the current count if the key exists; otherwise it starts from 0.
Each occurrence increases the count by 1

```python

# UCS420 Assignment 2
# Question 3 - Random Numbers
# Name : Hazel Latka
# Roll No : 1024160105
# Branch : COSE


# Importing the random module
import random

# Setting the random seed using the roll number
random.seed(1024160105)

print("Random seed has been set successfully.")
```

```python
# Generating a list of 100 random integers between 1 and 100

random_numbers = [random.randint(1, 100) for i in range(100)]

print("Random Numbers:")
print(random_numbers)
```

```python
# Finding the largest and smallest numbers

maximum = max(random_numbers)
minimum = min(random_numbers)

print("Maximum Number =", maximum)
print("Minimum Number =", minimum)
```

```python
# Creating a dictionary to store frequency

frequency = {}

for number in random_numbers:
    frequency[number] = frequency.get(number, 0) + 1

print("Frequency of each number:\n")

for key in sorted(frequency):
    print(key, ":", frequency[key])
```

```python
# Finding duplicate numbers

duplicates = []

for key, value in frequency.items():
    if value > 1:
        duplicates.append(key)

print("Numbers appearing more than once:")
print(sorted(duplicates))
```

```python
# Sorting the frequency dictionary in descending order

sorted_frequency = sorted(frequency.items(),
                          key=lambda item: item[1],
                          reverse=True)

print("Three Most Frequent Numbers:\n")

for number, count in sorted_frequency[:3]:
    print(number, "appears", count, "times")
```

Explanation
random.seed()
random.seed(1024160105)
Initializes the random number generator.
Using the same seed always produces the same sequence of random numbers.
This makes the results reproducible.
random.randint(a, b)
random.randint(1, 100)

Generates a random integer between 1 and 100 (both inclusive).

Dictionary Frequency Counting
frequency[number] = frequency.get(number, 0) + 1
If the number is already present in the dictionary, its count is increased by 1.
Otherwise, it starts with a count of 0 and becomes 1.
sorted() with lambda
sorted(frequency.items(),
       key=lambda item: item[1],
       reverse=True)
frequency.items() returns (number, count) pairs.
key=lambda item: item[1] tells Python to sort based on the frequency (count).
reverse=True sorts in descending order.

```python

# UCS420 Assignment 2
# Question 4 - Sets
# Name : Hazel Latka
# Roll No : 1024160105
# Branch : COSE


# Creating Set A using unique digits from the roll number

roll_no = "1024160105"

A = set(int(digit) for digit in roll_no)

# Creating Set B

B = {1, 2, 3, 4, 5}

print("Set A =", A)
print("Set B =", B)
```

```python
# Finding the union of A and B

union_set = A.union(B)

print("Union of A and B:")
print(union_set)
```

```python
# Finding common elements

intersection_set = A.intersection(B)

print("Intersection of A and B:")
print(intersection_set)
```

```python
# Elements present in A but not in B

difference_set = A.difference(B)

print("Difference (A - B):")
print(difference_set)
```

```python
# Elements present in either A or B but not in both

symmetric_difference = A.symmetric_difference(B)

print("Symmetric Difference:")
print(symmetric_difference)
```

```python
# Checking subset relationship

is_subset = A.issubset(B)

print("Is A a subset of B?")
print(is_subset)
```

```python
# Checking superset relationship

is_superset = B.issuperset(A)

print("Is B a superset of A?")
print(is_superset)
```

Explanation
Set

A set is an unordered collection of unique elements.

Example:

A = {1, 2, 3}

Duplicate values are automatically removed.

union()
A.union(B)

Returns all unique elements from both sets.

Example:

A = {1,2}
B = {2,3}

Union = {1,2,3}
intersection()
A.intersection(B)

Returns only the common elements.

Example:

A = {1,2,3}
B = {2,3,4}

Intersection = {2,3}
difference()
A.difference(B)

Returns elements that are in A but not in B.

Example:

A = {1,2,3}
B = {2,3}

Difference = {1}
symmetric_difference()
A.symmetric_difference(B)

Returns elements that belong to either set, but not both.

Example:

A = {1,2,3}
B = {2,3,4}

Symmetric Difference = {1,4}
issubset()
A.issubset(B)

Returns True if every element of A is present in B.

issuperset()
B.issuperset(A)

Returns True if B contains all elements of A.

```python

# UCS420 Assignment 2
# Question 5 - Dictionaries
# Name : Hazel Latka
# Roll No : 1024160105
# Branch : COSE


# Creating the student dictionary

student = {
    "Name": "Hazel Latka",
    "Roll Number": "1024160105",
    "Branch": "COSE",
    "Age": 20,
    "Home City": "Chandigarh",
    "CGPA": "Enter your CGPA here"
}

print("Student Dictionary:\n")
print(student)
```

```python
# Adding a new key-value pair

student["Email"] = "hazel.latka@example.com"

print("Dictionary after adding Email:\n")
print(student)
```

```python
# Updating the Home City

student["Home City"] = "Mohali"

print("Dictionary after updating Home City:\n")
print(student)
```

```python
# Removing the Email key

removed_value = student.pop("Email")

print("Removed Value:", removed_value)

print("\nDictionary after removing Email:\n")
print(student)
```

```python
# Displaying all keys

print("Dictionary Keys:\n")

for key in student.keys():
    print(key)
```

```python
# Displaying all values

print("Dictionary Values:\n")

for value in student.values():
    print(value)
```

```python
# Searching for a key

search_key = "Branch"

if search_key in student:
    print(search_key, "exists in the dictionary.")
    print("Value =", student[search_key])
else:
    print(search_key, "does not exist.")
```

```python
# Creating another student's dictionary

friend = {
    "Name": "Aman",
    "Roll Number": "1024160110",
    "Branch": "COSE",
    "Age": 20,
    "Home City": "Chandigarh",
    "CGPA": "8.40"
}

print("Friend Dictionary:\n")
print(friend)
```

```python
# Merging both dictionaries

combined_students = {
    "Student 1": student,
    "Student 2": friend
}

print("Merged Dictionary:\n")
print(combined_students)
```

Explanation
Dictionary

A dictionary stores data as key–value pairs.

Example:

student = {
    "Name": "Hazel Latka",
    "Age": 20
}
Adding a New Item
student["Email"] = "hazel.latka@example.com"

Adds a new key-value pair.

Updating a Value
student["Home City"] = "Mohali"

If the key already exists, its value is updated.

Removing an Item
student.pop("Email")

Removes the specified key and returns its value.

Displaying Keys
student.keys()

Returns all keys in the dictionary.

Displaying Values
student.values()

Returns all values stored in the dictionary.

Searching for a Key
if "Branch" in student:
    print(student["Branch"])

Checks whether a key exists before accessing it.

Merging Dictionaries
combined_students = {
    "Student 1": student,
