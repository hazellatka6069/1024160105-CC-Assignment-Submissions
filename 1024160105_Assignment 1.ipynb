```python
# ==========================================================
# PROGRAM 11 : DATA STRUCTURE - LIST
# ==========================================================

# Lists are ordered, mutable collections in Python.
# They can store different data types together.

# ==========================================================
# PROGRAM 11.1 : List Declaration
# ==========================================================

print("\n==============================")
print("11.1 List Declaration")
print("==============================")

# Creating a list with different data types

L = ["Pratham", "Sharma", 3.14, 3]

print("Original List :", L)

print("Number of Elements :", len(L))

# Explanation:
# [] creates a list.
# len() returns the number of elements.



# ==========================================================
# PROGRAM 11.2 : List Traversal using While Loop
# ==========================================================

print("\n==============================")
print("11.2 List Traversal (While)")
print("==============================")

L = ["Pratham", "Sharma", 3.14, 3]

print("Original List :", L)

i = 0

while i < len(L):

    print(L[i])

    i += 1

# Explanation:
# i starts from 0.
# Loop continues until last index.
# L[i] accesses element at index i.



# ==========================================================
# PROGRAM 11.3 : List Traversal using For Loop
# ==========================================================

print("\n==============================")
print("11.3 List Traversal (For)")
print("==============================")

L = ["Pratham", "Sharma", 3.14, 3]

print("Original List :", L)

for i in range(len(L)):

    print(L[i])

# Explanation:
# range(len(L)) generates list indices.
# for loop visits every index.



# ==========================================================
# PROGRAM 11.4 : For-Each Loop
# ==========================================================

print("\n==============================")
print("11.4 For Each Loop")
print("==============================")

L = ["Pratham", "Sharma", 3.14, 3]

print("Original List :", L)

for item in L:

    print(item)

# Explanation:
# Directly accesses elements.
# No need of index variable.



# ==========================================================
# PROGRAM 11.5 : Add and Delete Elements
# ==========================================================

print("\n==============================")
print("11.5 Add/Delete Elements")
print("==============================")

L = ["Pratham", "Sharma", 3.14, 3]

print("Original List :", L)

# append() adds element at end

L.append("Rahul")

print("After append() :", L)

# del removes element using index

del L[1]

print("After del :", L)

# Explanation:
#
# append() → Add new element
#
# del → Delete element using index
#
# List is mutable, so original list changes.
```

```python
# ==========================================================
# PROGRAM 14 : DATA STRUCTURE - SET
# ==========================================================

# A Set is an unordered collection of UNIQUE elements.
# Duplicate values are automatically removed.
# Sets are mutable (can be modified).

# Syntax:
# s = {1,2,3}
# or
# s = set([1,2,3])



# ==========================================================
# PROGRAM 14.1 : Declare Set
# ==========================================================

print("\n==============================")
print("14.1 Declare Set")
print("==============================")

# Duplicate elements are automatically removed

s = set(['A','B','E','F','E','F'])

print("Original Set =", s)

print("Number of Elements =", len(s))

# ----------------------------------------------------------
# Explanation
# ----------------------------------------------------------
#
# Set stores only UNIQUE elements.
#
# Duplicate values are automatically removed.
#
# Order of elements may change because
# Set is unordered.
#
# ----------------------------------------------------------



# ==========================================================
# PROGRAM 14.2 : Set Operations
# ==========================================================

print("\n==============================")
print("14.2 Set Operations")
print("==============================")

a = set(['A','B','E','F'])

b = set(['A','C','D','E'])

print("Set A =", a)

print("Set B =", b)

# Union

print("\nUnion")

print(a.union(b))

# Intersection

print("\nIntersection")

print(a.intersection(b))

# Difference

print("\nDifference A - B")

print(a - b)

print(a.difference(b))

print("\nDifference B - A")

print(b - a)

print(b.difference(a))

# Symmetric Difference

print("\nSymmetric Difference")

print(a.symmetric_difference(b))

print(b.symmetric_difference(a))

# ----------------------------------------------------------
# Explanation
# ----------------------------------------------------------
#
# Union
# All elements from both sets.
#
# Intersection
# Common elements.
#
# Difference
# Elements present in first set only.
#
# Symmetric Difference
# Elements present in either set
# but NOT common.
#
# ----------------------------------------------------------



# ==========================================================
# PROGRAM 14.3 : Add, Remove and Pop
# ==========================================================

print("\n==============================")
print("14.3 Add, Remove and Pop")
print("==============================")

a = set(['A','B','E','F'])

print("Original Set")

print(a)

# Add Element

a.add("D")

print("\nAfter Adding D")

print(a)

# Duplicate not added

a.add("D")

print("\nAdding D Again")

print(a)

# Remove Element

a.remove("D")

print("\nAfter Removing D")

print(a)

# pop() removes a random element

removed = a.pop()

print("\nRemoved Element =", removed)

print(a)

removed = a.pop()

print("\nRemoved Element =", removed)

print(a)

# ----------------------------------------------------------
# Explanation
# ----------------------------------------------------------
#
# add()
# Adds new element.
#
# Duplicate values are ignored.
#
# remove()
# Removes specified element.
#
# pop()
# Removes a RANDOM element.
#
# Since set is unordered,
# pop() does NOT remove the last element.
#
# ----------------------------------------------------------



# ==========================================================
# SUMMARY OF SET
# ==========================================================

# set()
# Create Set
#
# add()
# Add Element
#
# remove()
# Remove Element
#
# pop()
# Remove Random Element
#
# union()
# Combine Sets
#
# intersection()
# Common Elements
#
# difference()
# Different Elements
#
# symmetric_difference()
# Non-common Elements
#
# Sets contain UNIQUE elements only.
#
# Sets are unordered.
#
# Sets are mutable.
```

```python
# ==========================================================
# PROGRAM 16 : FILE HANDLING
# ==========================================================

# File Handling is used to store data permanently in files.
#
# Common File Modes:
#
# 'r'  -> Read Mode
# 'w'  -> Write Mode (Creates new file or overwrites existing file)
# 'a'  -> Append Mode (Adds data at the end)
#
# open()  -> Opens a file
# write() -> Writes into file
# read()  -> Reads entire file
# close() -> Closes file



# ==========================================================
# PROGRAM 16.1 : Write Numbers 1 to 10 into a File
# ==========================================================

print("\n==============================")
print("16.1 Write into File")
print("==============================")

# Open file in write mode
# If file does not exist, Python creates it.

fp = open("result.txt", "w")

# Write numbers 1 to 10

for i in range(1, 11):

    fp.write(str(i) + "\n")

# Close file

fp.close()

print("Writing Completed!")

print("Open result.txt to see the contents.")

# ----------------------------------------------------------
# Explanation
# ----------------------------------------------------------
#
# open(filename,'w')
#
# Opens file in write mode.
#
# Existing contents are erased.
#
# write()
#
# Writes text into file.
#
# str()
#
# Converts integer into string.
#
# "\n"
#
# Moves to next line.
#
# close()
#
# Saves and closes the file.
#
# ----------------------------------------------------------



# ==========================================================
# PROGRAM 16.2 : Read File and Display Contents
# ==========================================================

print("\n==============================")
print("16.2 Read File")
print("==============================")

# Open file in read mode

fp = open("result.txt", "r")

# Read line by line

for line in fp:

    print(line.strip())

# Close file

fp.close()

# ----------------------------------------------------------
# Explanation
# ----------------------------------------------------------
#
# 'r'
#
# Opens file in read mode.
#
# for line in file
#
# Reads one line at a time.
#
# strip()
#
# Removes extra spaces and newline (\n).
#
# ----------------------------------------------------------



# ==========================================================
# PROGRAM 16.3 : Read One File and Write into Another File
# ==========================================================

print("\n==============================")
print("16.3 Copy File in Uppercase")
print("==============================")

# Open source file

Readfp = open("result.txt", "r")

# Open destination file

Writefp = open("abc.txt", "w")

# Read each line and convert into uppercase

for line in Readfp:

    Writefp.write(line.upper())

# Close both files

Readfp.close()

Writefp.close()

print("Uppercase File Created Successfully!")

print("Open abc.txt to view the contents.")

# ----------------------------------------------------------
# Explanation
# ----------------------------------------------------------
#
# line.upper()
#
# Converts lowercase letters into uppercase.
#
# write()
#
# Stores modified text into another file.
#
# Source file remains unchanged.
#
# Destination file contains updated data.
#
# ----------------------------------------------------------



# ==========================================================
# BONUS PROGRAM 16.4 : Read Entire File at Once
# ==========================================================

print("\n==============================")
print("16.4 Read Entire File")
print("==============================")

fp = open("result.txt", "r")

content = fp.read()

print(content)

fp.close()

# ----------------------------------------------------------
# Explanation
# ----------------------------------------------------------
#
# read()
#
# Reads complete file at once.
#
# Best for small files.
#
# ----------------------------------------------------------



# ==========================================================
# BONUS PROGRAM 16.5 : Append Data to Existing File
# ==========================================================

print("\n==============================")
print("16.5 Append to File")
print("==============================")

fp = open("result.txt", "a")

fp.write("11\n")

fp.write("12\n")

fp.close()

print("Data Appended Successfully!")

# ----------------------------------------------------------
# Explanation
# ----------------------------------------------------------
#
# 'a'
#
# Append mode.
#
# Existing data is preserved.
#
# New data is added at the end.
#
# ----------------------------------------------------------



# ==========================================================
# BONUS PROGRAM 16.6 : Count Number of Lines in a File
# ==========================================================

print("\n==============================")
print("16.6 Count Lines")
print("==============================")

fp = open("result.txt", "r")

count = 0

for line in fp:

    count += 1

fp.close()

print("Total Lines =", count)

# ----------------------------------------------------------
# Explanation
# ----------------------------------------------------------
#
# Each iteration reads one line.
#
# Counter keeps track of total lines.
#
# ----------------------------------------------------------



# ==========================================================
# SUMMARY OF FILE HANDLING
# ==========================================================

# open(file,'r')
# Open File for Reading
#
# open(file,'w')
# Open File for Writing
#
# open(file,'a')
# Open File for Appending
#
# write()
# Write Data into File
#
# read()
# Read Entire File
#
# for line in file
# Read Line by Line
#
# upper()
# Convert Text to Uppercase
#
# strip()
# Remove Leading/Trailing Spaces and '\n'
#
# close()
# Close File
#
# Always close the file after reading or writing.
#
# File Handling is used for permanent storage of data.
```
