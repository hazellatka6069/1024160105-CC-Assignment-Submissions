```python

# UCS420 - COGNITIVE COMPUTING
# ASSIGNMENT 4
# A Cognitive FAQ System Using Pandas


# NAME       :Hazel Latka
# ROLL NO.   : 1024160105
#
# IMPORTANT CONCEPTS USED:
# - Pandas DataFrame
# - Lists and dictionaries
# - Functions
# - String manipulation
# - Sets
# - Keyword matching
# - DataFrame filtering
# - DataFrame indexing
# - groupby()
# - CSV file handling



import pandas as pd



# Q1. BUILD YOUR PERSONALIZED KNOWLEDGE BASE


# The assignment requires exactly 6 FAQ entries:
#   4 fixed entries given in the question
#   + 2 personalized entries based on the LAST TWO digits
#     of our roll number.

# Our roll number = 1024160105
# Last two digits = 05

# Therefore:

# Digit 0:
# 0 % 3 = 0
# category[0] = "billing"

# Digit 5:
# 5 % 3 = 2
# category[2] = "general"

# So our personalized categories are:
# 1. billing
# 2. general
# The assignment says to create a realistic question,
# answer and 3 keywords for each personalized category.



# Store the roll number.
ROLL_NUMBER = 1024160105


# Extract the last two digits.

# % 100 gives the remainder after dividing by 100,
# which gives the last two digits.

# Example:
# 1024160105 % 100 = 5

# zfill(2) changes "5" into "05".
# This is important because our first digit is actually 0.
last_two_digits = str(ROLL_NUMBER % 100).zfill(2)


# Extract each digit separately.
digit1 = int(last_two_digits[0])
digit2 = int(last_two_digits[1])


# The assignment provides this category mapping:

# ["billing", "account", "general"][d % 3]

# Therefore:
# remainder 0 -> billing
# remainder 1 -> account
# remainder 2 -> general

categories = ["billing", "account", "general"]


# Find the category for each of our two digits.
category1 = categories[digit1 % 3]
category2 = categories[digit2 % 3]


# Print the calculation so that we can understand and
# demonstrate it during viva if asked.

print("================================================")
print("Q1 - ROLL NUMBER CATEGORY CALCULATION")
print("================================================")

print("Roll Number:", ROLL_NUMBER)
print("Last two digits:", last_two_digits)

print("\nFirst digit:", digit1)
print("Calculation:", digit1, "% 3 =", digit1 % 3)
print("Category:", category1)

print("\nSecond digit:", digit2)
print("Calculation:", digit2, "% 3 =", digit2 % 3)
print("Category:", category2)



# FIXED FAQ ENTRIES


# These four entries are given directly in the assignment.
# A dictionary stores data in key-value pairs.
# For example:

# "question" -> question text
# "answer"   -> answer text
# "keywords" -> keywords used for matching
# "category" -> FAQ category

# A list of dictionaries is used because each dictionary
# represents one FAQ row.


fixed_entries = [

    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },

    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },

    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },

    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]



# PERSONALIZED FAQ ENTRY 1


# Our first digit is 0.

# 0 % 3 = 0
# category[0] = billing

# Therefore this entry belongs to the billing category.



personalized_entry1 = {

    "question": "how do i check my billing details",

    "answer": "You can check your billing details from the billing section of your account.",

    # Exactly 3 keywords are provided for this personalized entry.
    "keywords": "billing invoice charges",

    "category": category1
}



# PERSONALIZED FAQ ENTRY 2


# Our second digit is 5.

# 5 % 3 = 2
# category[2] = general

# Therefore this entry belongs to the general category.



personalized_entry2 = {

    "question": "how can i get general information",

    "answer": "You can find general information in the help and information section.",

    # Exactly 3 keywords are provided.
    "keywords": "general information help",

    "category": category2
}



# COMBINE ALL 6 ENTRIES
# We combine:
# 4 fixed entries
# +
# 2 personalized entries
# =
# 6 total entries



all_entries = fixed_entries + [
    personalized_entry1,
    personalized_entry2
]


# Create a Pandas DataFrame.

# A DataFrame is a two-dimensional table consisting of
# rows and columns.

# Each dictionary becomes one row.
# Dictionary keys become column names.

df = pd.DataFrame(all_entries)



# PRINT FINAL 6-ROW DATAFRAME


print("\n================================================")
print("Q1 - FINAL 6-ROW FAQ DATAFRAME")
print("================================================")

print(df)


# VIVA POINT:
# What is Pandas?
# Pandas is a Python library used for data manipulation
# and data analysis.
#
# VIVA POINT:
# What is a DataFrame?
# A DataFrame is a two-dimensional labelled data structure
# consisting of rows and columns.
#
# VIVA POINT:
# Why did we use a list of dictionaries?
# Each dictionary represents one FAQ entry, while the list
# stores all FAQ entries together.
#
# VIVA POINT:
# Why did we use zfill(2)?
# Because our last two digits are 05.
# Without zfill(), Python would represent it as "5".
# zfill(2) ensures that it remains "05", allowing us to
# correctly extract 0 and 5.



# Q2. GENERATE AND SCORE A HYPOTHESIS


# The assignment asks us to implement a scoring function
# that takes a query string and returns matching FAQ entries
# ranked by confidence.
#
# Our simple scoring method:

# 1. Convert the query into lowercase.
# 2. Split the query into words.
# 3. Convert the FAQ keywords into lowercase.
# 4. Split keywords into words.
# 5. Find common words between query and keywords.
# 6. Count the common words.
# 7. Use that count as the confidence score.
# 8. Sort entries from highest score to lowest score.

# This is a simple keyword-based matching approach.
# It is NOT a machine-learning model.



def score_query(query, df):

    # Convert the query to lowercase.

    # lower() makes matching case-insensitive.

    # Example:
    # "Fee" becomes "fee"
    # "FEE" becomes "fee"

    query = query.lower()


    # split() divides a string into individual words.

    # Example:
    # "how can i pay fee"
    #
    # becomes:
    # ["how", "can", "i", "pay", "fee"]
    #
    # set() removes duplicate words and makes set operations
    # such as intersection() easy.

    query_words = set(query.split())


    # Create an empty list to store results.
    results = []


    # iterrows() allows us to process the DataFrame one row
    # at a time.

    # index = row index
    # row   = data contained in that row

    for index, row in df.iterrows():

        # Convert the keywords of the current FAQ entry
        # into a set of lowercase words.

        keyword_words = set(
            row["keywords"].lower().split()
        )


        # intersection() finds words that occur in BOTH sets.

        # Example:

        # query words:
        # {"pay", "fee"}

        # keywords:
        # {"pay", "payment", "upi", "fee"}

        # intersection:
        # {"pay", "fee"}

        matching_words = query_words.intersection(
            keyword_words
        )


        # The number of matching words becomes the score.

        # len() returns the number of elements in the set.

        score = len(matching_words)


        # Store the result as a dictionary.

        results.append({

            "question": row["question"],

            "answer": row["answer"],

            "category": row["category"],

            "score": score,

            "matching_keywords": ", ".join(matching_words)
        })


    # Convert the list of result dictionaries into a DataFrame.

    result_df = pd.DataFrame(results)


    # Sort results according to score.

    # ascending=False means highest score appears first.

    result_df = result_df.sort_values(
        by="score",
        ascending=False
    )


    # reset_index(drop=True) creates a fresh index after sorting.

    # drop=True prevents the old index from becoming
    # an additional column.

    result_df = result_df.reset_index(drop=True)


    # Return the ranked results to the calling code.

    return result_df



# TEST Q2


print("\n================================================")
print("Q2 - QUERY SCORING")
print("================================================")


# Ask the user to enter a query.

user_query = input("Enter your FAQ query: ")


# Call our scoring function.

scored_results = score_query(
    user_query,
    df
)


# Print the ranked results.

print("\nRanked FAQ Results:")
print(scored_results)


# VIVA POINT:
# What is a function?
# A function is a reusable block of code designed to perform
# a particular task.

# VIVA POINT:
# Why did we use a function here?
# To make the scoring logic reusable for different queries.

# VIVA POINT:
# What does split() do?
# It divides a string into a list of words.

# VIVA POINT:
# What does set() do?
# It creates a set containing unique elements.

# VIVA POINT:
# What does intersection() do?
# It returns elements common to both sets.

# VIVA POINT:
# How is confidence calculated?
# Confidence is represented by the number of keywords that
# match between the user's query and an FAQ entry.

# VIVA POINT:
# Why is ascending=False used?
# Because we want the highest score at the top.

# VIVA POINT:
# Is this semantic understanding?
# No. This is keyword-based matching. It checks words,
# rather than understanding the actual meaning of a sentence.



# Q3. same_category(category_name, df)


# The assignment asks us to write a function:

# same_category(category_name, df)

# It should return all questions belonging to a particular
# category.

# We will use the category of our FIRST personalized entry.
# Our first personalized category is "billing".



def same_category(category_name, df):

    # Convert category name to lowercase.

    # This makes our comparison case-insensitive.

    category_name = category_name.lower()


    # Boolean filtering:

    # df["category"] == category_name

    # creates a True/False value for every row.

    # Rows with True are selected.

    return df[
        df["category"].str.lower() == category_name
    ]


# CALL FUNCTION USING PERSONALIZED ENTRY CATEGORY


print("\n================================================")
print("Q3 - SAME CATEGORY")
print("================================================")


# Our first personalized entry has category "billing".

category_to_search = category1


print("Category searched:", category_to_search)


# Call the function.

category_result = same_category(
    category_to_search,
    df
)


print("\nFAQ entries belonging to this category:")
print(category_result)


# VIVA POINT:
# What is Boolean filtering?
# Boolean filtering means selecting rows of a DataFrame
# according to a condition.

# Example:

# df[df["category"] == "billing"]

# selects only rows whose category is billing.

# VIVA POINT:
# What does .str.lower() do?
# It converts every string value in the selected Pandas
# column to lowercase.

# VIVA POINT:
# Why did we pass df to the function?
# Because the function needs the DataFrame from which it
# has to select the matching category entries.



# Q4. ADD A NEW KEYWORD AND SAVE AS CSV


# The assignment asks us to:

# 1. Pick any one FAQ entry.
# 2. Ask the user for a new keyword.
# 3. Add the keyword to that entry.
# 4. Save the complete updated DataFrame.

# Required filename:

# <your_roll_number>_faq_data.csv

# Therefore our filename will be:

# 1024160105_faq_data.csv



print("\n================================================")
print("Q4 - ADD NEW KEYWORD")
print("================================================")


# We choose the first FAQ entry.

# DataFrame indexing starts from 0.
# Therefore index 0 means the first row.

selected_index = 0


print("Selected FAQ entry:")
print(df.loc[selected_index])


# Ask the user to enter a new keyword.

new_keyword = input(
    "\nEnter a new keyword to add to this FAQ: "
).strip()


# Check that the user did not enter an empty string.

if new_keyword:

    # Add the new keyword to the existing keywords.

    # Example:

    # Existing:
    # fee cost price charge

    # New keyword:
    # money

    # Updated:
    # fee cost price charge money

    df.loc[selected_index, "keywords"] = (
        df.loc[selected_index, "keywords"]
        + " "
        + new_keyword
    )



# SAVE UPDATED DATAFRAME TO CSV


# Create the required filename using the roll number.

filename = f"{ROLL_NUMBER}_faq_data.csv"


# to_csv() saves the DataFrame as a CSV file.

# index=False prevents Pandas from saving the DataFrame's
# numerical index as an additional column.

df.to_csv(
    filename,
    index=False
)


print("\nUpdated FAQ entry:")
print(df.loc[selected_index])

print("\nComplete updated DataFrame:")
print(df)

print("\nDataFrame successfully saved as:")
print(filename)


# VIVA POINT:
# What is CSV?
# CSV stands for Comma-Separated Values.
# It is a common file format for storing tabular data.

# VIVA POINT:
# What does to_csv() do?
# It writes a Pandas DataFrame into a CSV file.

# VIVA POINT:
# Why use index=False?
# Without index=False, Pandas may save the DataFrame index
# as an extra column in the CSV file.

# VIVA POINT:
# What does .loc[] do?
# .loc[] is used for label-based selection and modification
# of rows and columns.

# VIVA POINT:
# Why did we use .strip()?
# It removes unnecessary whitespace from the beginning
# and end of the user's input.



# Q5. COUNT FAQ ENTRIES PER CATEGORY USING groupby()


# The assignment specifically asks us to use groupby().

# We want to know how many FAQ entries belong to:
# billing
# account
# general



print("\n================================================")
print("Q5 - FAQ COUNT PER CATEGORY")
print("================================================")


# groupby("category") groups all rows having the same
# category value.

# size() counts the number of rows in every group.

category_counts = df.groupby("category").size()


print("Number of FAQ entries in each category:")
print(category_counts)


# VIVA POINT:
# What is groupby()?
# groupby() is used to divide data into groups based on
# the values of a column.

# VIVA POINT:
# What does size() do after groupby()?
# It counts the number of rows in each group.

# VIVA POINT:
# Difference between groupby() and filtering?
# Filtering selects rows satisfying a condition.
# groupby() divides the data into groups so that we can
# perform calculations or analysis on each group.



# Q6. HANDLE TIES IN HIGHEST SCORE


# Q2 ranked the FAQ entries according to their scores.
#
# However, if two or more entries have exactly the same
# highest score, we should display ALL of them.

# We therefore modify the scoring function.

# Steps:

# 1. Calculate score for every FAQ entry.
# 2. Find the highest score using max().
# 3. Select EVERY row whose score equals that highest score.

# This prevents the program from silently selecting only
# one FAQ entry when there is a tie.


def score_query_with_ties(query, df):

    # Convert the query into lowercase.

    query = query.lower()


    # Convert query into a set of words.

    query_words = set(
        query.split()
    )


    # Store results for every FAQ entry.

    results = []


    # Process each FAQ entry.

    for index, row in df.iterrows():

        # Convert FAQ keywords into a set.

        keyword_words = set(
            row["keywords"].lower().split()
        )


        # Find common words between query and keywords.

        matching_words = query_words.intersection(
            keyword_words
        )


        # Count common words to get score.

        score = len(matching_words)


        # Store result.

        results.append({

            "question": row["question"],

            "answer": row["answer"],

            "category": row["category"],

            "score": score,

            "matching_keywords": ", ".join(matching_words)
        })


    # Convert results into a DataFrame.

    result_df = pd.DataFrame(results)


    # Find the maximum score.
    #
    # max() returns the highest score present in the
    # score column.

    highest_score = result_df["score"].max()


    # Select ALL entries whose score equals the highest score.
        # This is the main difference between Q2 and Q6.

    # We are NOT using:

    # result_df.iloc[0]

    # because that would return only one result.

    # Instead, we use Boolean filtering to return every
    # equally good result.

    tied_results = result_df[
        result_df["score"] == highest_score
    ]


    # Reset the index for clean output.

    tied_results = tied_results.reset_index(
        drop=True
    )


    # Return all highest-scoring entries.

    return tied_results


# Q6 DEMONSTRATION 1 - QUERY THAT PRODUCES A TIE


# We use "fee".

# The two fixed billing entries are:

# "what is the annual fee"
# "how can i pay the fee"

# Both contain "fee" in their keywords.

# Therefore both receive the same score of 1.

# The modified function should display BOTH entries.



print("\n================================================")
print("Q6 - TIE DEMONSTRATION")
print("================================================")


tie_query = "fee"


tie_results = score_query_with_ties(
    tie_query,
    df
)


print("Query:", tie_query)

print("\nAll entries having the highest score:")
print(tie_results)



# Q6 DEMONSTRATION 2 - QUERY THAT DOES NOT PRODUCE A TIE

# We use "password".

# The password FAQ has the keyword "password".
# Therefore it should get the highest score.



non_tie_query = "password"


non_tie_results = score_query_with_ties(
    non_tie_query,
    df
)


print("\n================================================")
print("Q6 - NON-TIE DEMONSTRATION")
print("================================================")


print("Query:", non_tie_query)

print("\nHighest-scoring entry/entries:")
print(non_tie_results)



# FINAL VIVA REVISION

# Q1:
# Why did we use Pandas?

# Answer:
# Pandas provides data structures such as DataFrame that
# make it easy to store, manipulate and analyse tabular data.


# Q1:
# Why exactly 6 rows?

# Answer:
# The assignment requires 4 fixed FAQ entries plus 2
# personalized entries based on the last two roll-number
# digits.


# Q1:
# How were our personalized categories calculated?

# Answer:
# Our roll number is 1024160105.
# Last two digits are 05.

# For digit 0:
# 0 % 3 = 0 -> billing

# For digit 5:
# 5 % 3 = 2 -> general


# Q2:
# What is the purpose of score_query()?

# Answer:
# It takes a user's query, compares its words with FAQ
# keywords, calculates a matching score, and ranks the
# FAQ entries according to that score.


# Q2:
# How is the score calculated?

# Answer:
# Score = number of common words between the query and
# the FAQ keywords.


# Q2:
# Why use lowercase?

# Answer:
# To make keyword matching case-insensitive.
# For example, "Fee" and "fee" should be treated as the
# same word.


# Q2:
# Why use set intersection?

# Answer:
# Set intersection returns the elements common to two sets.
# It allows us to find the keywords shared by the query
# and an FAQ entry.


# Q3:
# What does same_category() return?

# Answer:
# It returns all DataFrame rows whose category matches
# the category supplied to the function.


# Q4:
# What is to_csv()?

# Answer:
# It saves the DataFrame in CSV format.


# Q4:
# Why index=False?

# Answer:
# It prevents the DataFrame index from being saved as
# an unnecessary additional column.


# Q5:
# Why use groupby()?

# Answer:
# groupby() groups FAQ entries according to category,
# allowing us to count how many entries belong to each
# category.


# Q6:
# What is a tie?

# Answer:
# A tie occurs when two or more FAQ entries have exactly
# the same highest confidence score.


# Q6:
# How do we handle ties?

# Answer:
# First we find the maximum score using max().
# Then we use Boolean filtering to select every row whose
# score is equal to that maximum score.


# Q6:
# Why don't we simply select the first row?

# Answer:
# Because that would hide other FAQ entries having the
# same highest score. The assignment specifically requires
# all tied highest-scoring entries to be displayed.


# GENERAL VIVA:

# What is a library?
# A library is a collection of pre-written code that can
# be imported and used in a program.

# What does import pandas as pd mean?
# It imports the Pandas library and gives it the shorter
# alias "pd".

# What is a dictionary?
# A dictionary stores data as key-value pairs.

# What is a list?
# A list is an ordered collection of elements.

# What is len()?
# len() returns the number of elements in an object.

# What is max()?
# max() returns the largest value from a collection or
# Series.

# What is reset_index(drop=True)?
# It creates a new sequential index and prevents the old
# index from being retained as a column.

# Is this FAQ system using AI/ML?
# The code in this assignment implements a simple
# rule-based keyword matching approach. It does not train
# a machine-learning model.


# END OF ASSIGNMENT 4
```
