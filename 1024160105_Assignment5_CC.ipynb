```python

# ASSIGNMENT - 4
# UCS420 - COGNITIVE COMPUTING
# NumPy - Introduction - 1
#
# Name    : Hazel Latka
# Roll No. : 1024160105


import numpy as np


# Q1. Create a NumPy 1-D array with 5 elements and perform:
#     a) Addition of 2 to all elements
#     b) Multiply 3 with all elements
#     c) Divide every element by 2


arr1 = np.array([10, 20, 30, 40, 50])

print("Q1 - Original Array:")
print(arr1)

print("\na) Addition of 2:")
print(arr1 + 2)

print("\nb) Multiplication by 3:")
print(arr1 * 3)

print("\nc) Division by 2:")
print(arr1 / 2)



# Q2(a). Reverse the NumPy array


arr = np.array([1, 2, 3, 6, 4, 5])

print("\n\nQ2(a) - Original Array:")
print(arr)

# [::-1] reverses the array.
print("Reversed Array:")
print(arr[::-1])



# Q2(b). Find the most frequent value and its indices

# i. x = [1,2,3,4,5,1,2,1,1,1]
# ii. y = [1,1,1,2,3,4,2,4,3,3]


x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

# np.unique(..., return_counts=True) gives:
# unique values and the number of times each value occurs.
values_x, counts_x = np.unique(x, return_counts=True)

# argmax() gives the position of the largest count.
most_frequent_x = values_x[np.argmax(counts_x)]

# np.where() gives the indices where the most frequent value occurs.
indices_x = np.where(x == most_frequent_x)[0]

print("\n\nQ2(b)(i) - Array x:")
print(x)

print("Most frequent value:", most_frequent_x)
print("Indices:", indices_x)


y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values_y, counts_y = np.unique(y, return_counts=True)

most_frequent_y = values_y[np.argmax(counts_y)]

indices_y = np.where(y == most_frequent_y)[0]

print("\nQ2(b)(ii) - Array y:")
print(y)

print("Most frequent value:", most_frequent_y)
print("Indices:", indices_y)



# Q3. Given 2-D array:

# [10 20 30]
# [40 50 60]
# [70 80 90]
# a) Access 1st row, 2nd column
# b) Access 3rd row, 1st column


arr2 = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("\n\nQ3 - 2-D Array:")
print(arr2)

# Python/NumPy uses zero-based indexing.
# Therefore, 1st row = index 0 and 2nd column = index 1.
print("\na) 1st row, 2nd column:")
print(arr2[0, 1])

# 3rd row = index 2 and 1st column = index 0.
print("\nb) 3rd row, 1st column:")
print(arr2[2, 0])



# Q4. Create a 1-D NumPy array named Hazel_Latka with evenly
#     spaced 25 numbers from 10 to 100 using linspace().

# Print:
# a) Dimensions
# b) Shape
# c) Total elements
# d) Data type
# e) Total bytes consumed
# f) Transpose using reshape()
# g) Check whether T attribute can also be used


Hazel_Latka = np.linspace(10, 100, 25)

print("\n\nQ4 - Hazel_Latka Array:")
print(Hazel_Latka)

# ndim gives the number of dimensions.
print("\na) Dimensions:")
print(Hazel_Latka.ndim)

# shape gives the size along each dimension.
print("\nb) Shape:")
print(Hazel_Latka.shape)

# size gives the total number of elements.
print("\nc) Total Elements:")
print(Hazel_Latka.size)

# dtype gives the data type of elements.
print("\nd) Data Type:")
print(Hazel_Latka.dtype)

# nbytes gives total memory occupied by the array elements.
print("\ne) Total Bytes Consumed:")
print(Hazel_Latka.nbytes)

# 25 elements can be reshaped into a 5 x 5 matrix.
# reshape() does not change the actual data; it changes its shape.
Hazel_Latka_reshaped = Hazel_Latka.reshape(5, 5)

print("\nf) Array after reshape(5,5):")
print(Hazel_Latka_reshaped)

# Transpose exchanges rows and columns.
# For a 5 x 5 matrix, the result is also 5 x 5.
Hazel_Latka_transpose = Hazel_Latka.reshape(5, 5).T

print("\nTranspose using reshape() and .T:")
print(Hazel_Latka_transpose)

# Yes, T can also be used directly on a NumPy array.
# However, for a 1-D array, .T does NOT change its shape.
# Therefore, we first reshape the 1-D array into 5 x 5.
Hazel_Latka_T = Hazel_Latka.reshape(5, 5).T

print("\nTranspose using T attribute:")
print(Hazel_Latka_T)


# Q5. Create a 2-D array of 3 rows and 4 columns named:
#     ucs420_Hazel_Latka
#
# Values:
# 10, 20, 30, 40
# 50, 60, 70, 80
# 90, 15, 20, 35
#
# Find:
# a) Mean
# b) Median
# c) Maximum
# d) Minimum
# e) Unique elements
# f) Reshape into 4 x 3
# g) Resize into 2 x 3


ucs420_Hazel_Latka = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("\n\nQ5 - Original 3 x 4 Array:")
print(ucs420_Hazel_Latka)

# Mean = arithmetic average of all elements.
print("\na) Mean:")
print(np.mean(ucs420_Hazel_Latka))

# Median = middle value after arranging the values in order.
print("\nb) Median:")
print(np.median(ucs420_Hazel_Latka))

# Maximum gives the largest element.
print("\nc) Maximum:")
print(np.max(ucs420_Hazel_Latka))

# Minimum gives the smallest element.
print("\nd) Minimum:")
print(np.min(ucs420_Hazel_Latka))

# unique() returns each distinct value only once.
print("\ne) Unique Elements:")
print(np.unique(ucs420_Hazel_Latka))



# Reshape 3 x 4 into 4 x 3.
#
# Both arrays contain 12 elements, so reshape is possible.


reshaped_ucs420_Hazel_Latka = ucs420_Hazel_Latka.reshape(4, 3)

print("\nf) Reshaped Array (4 x 3):")
print(reshaped_ucs420_Hazel_Latka)


# Resize 3 x 4 into 2 x 3.
#
# resize() changes the total number of elements.
# Here the new array requires only 6 elements.

resized_ucs420_Hazel_Latka = np.resize(
    ucs420_Hazel_Latka,
    (2, 3)
)

print("\ng) Resized Array (2 x 3):")
print(resized_ucs420_Hazel_Latka)


```
