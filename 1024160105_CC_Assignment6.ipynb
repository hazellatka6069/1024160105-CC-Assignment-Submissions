```python
import numpy as np
temperature=np.array([25,28,31,35,38,27,33,40])
corrected=temperature+2
print(corrected)
fahrenheit=(temperature*9/5)+32
print(fahrenheit)
greater=temperature[temperature>32]
print(greater)
print(np.sum(temperature>32))
```

```python
import numpy as np
steps=np.array([[5000,6200,7100],
[8000,7500,9000],
[4500,5100,4800],
[9000,8500,9500]])
print(np.sum(steps))
print(np.mean(steps))
print(np.max(steps))
print(np.min(steps))
print(np.sum(steps,axis=0))
print(np.sum(steps,axis=1))
position=np.unravel_index(np.argmax(steps),steps.shape)
print(position)
```

```python
import numpy as np
original=np.array([1,2,3,4,5,6])
subset=original[1:5]
print(subset)
subset[0]=999
print(original)
print(subset)
copied=original[1:5].copy()
copied[0]=500
print(original)
print(copied)
arr=np.arange(1,13).reshape(3,4)
print(arr)
print(arr[0])
print(arr[-1])
print(arr[:,1])
print(arr[1:3,2:4])
flat=arr.flatten()
rav=arr.ravel()
print(flat)
print(rav)
rav[0]=100
print(arr)
flat[1]=200
print(arr)
print(arr.shape)
print(arr.ndim)
print(arr.size)
print(arr.dtype)
```

```python
import numpy as np
X=np.array([[6,70,3],
[5,50,6],
[8,80,2],
[4,30,8]])
y=np.array([40,65,30,85])
print(X.shape)
print(X.ndim)
XT=X.T
print(XT)
XTX=XT@X
print(XTX)
inverse=np.linalg.inv(XTX)
print(inverse)
beta=inverse@XT@y
print(beta)
new_user=np.array([5,40,7])
prediction=new_user@beta
print(prediction)
```
