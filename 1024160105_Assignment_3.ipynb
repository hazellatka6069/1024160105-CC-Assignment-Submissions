```python

# Q1. CREATE A DATASET USING PANDAS


import pandas as pd

# Creating a dictionary containing the dataset
data = {
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward",
             "Frank", "Grace", "Harry", "Isha", "Jack"],

    "Age": [29, 34, 41, 28, 38, 25, 31, 45, 27, 36],

    "Department": ["HR", "IT", "IT", "Marketing", "Sales",
                   "HR", "Marketing", "IT", "Sales", "HR"],

    "Salary": [50000, 70000, 65000, 55000, 60000,
               48000, 58000, 80000, 52000, 62000]
}

# Convert dictionary into DataFrame
df = pd.DataFrame(data)

# Display the DataFrame
print(df)
```

Pandas is a Python library used for data manipulation and analysis.
DataFrame is a two-dimensional tabular data structure in Pandas.
pd.DataFrame() converts data into a DataFrame.
A DataFrame has rows and columns.

```python

# Q2. LOCATE ROWS 0, 4, 7 AND 8


# iloc is used for integer-position based indexing

selected_rows = df.iloc[[0, 4, 7, 8]]

print(selected_rows)
df.iloc[[0, 4, 7, 8]]
```


Select rows at positions 0, 4, 7 and 8.

Q: What is iloc?
iloc is used for integer-position based indexing.

```python

# Q3.1 SELECT ROWS FROM INDEX 3 TO 7


# iloc uses position.
# 3:8 is used because the ending position in iloc is excluded.

result = df.iloc[3:8]

print(result)
```

Remember

iloc[3:8] gives:

3, 4, 5, 6, 7

because the ending index 8 is excluded.

```python

# Q3.2 SELECT ROWS 4 TO 8 AND COLUMNS 2 TO 4

# iloc uses zero-based indexing.
# Rows 4,5,6,7,8  -> 4:9
# Columns 2,3,4   -> 2:5
# Ending index is excluded in Python.

result = df.iloc[4:9, 2:5]

print(result)
```

```python

# Q3.3 SELECT ALL ROWS AND COLUMNS 1 TO 3


# All rows -> :
# Columns 1,2,3 -> 1:4

result = df.iloc[:, 1:4]

print(result)
```

: means all rows.

```python
df.iloc[:, 1:4]
```

means:

Select all rows and columns 1, 2 and 3.

```python
# Q4. READ CSV FILE AND DISPLAY FIRST 5 ROWS

import pandas as pd

# Download Iris.csv directly from GitHub
url = "https://raw.githubusercontent.com/uiuc-cse/data-fa14/gh-pages/data/iris.csv"

# Read CSV file
df_iris = pd.read_csv(url)

# Display first 5 rows
print(df_iris.head())
```

```python

# Q5. DELETE ROW 4 AND COLUMN 3


# Make a copy so that the original DataFrame remains unchanged
df_modified = df_iris.copy()

# Delete row with index 4
df_modified = df_modified.drop(index=4)

# Delete column with index 3
# First find the column name at position 3

column_name = df_modified.columns[3]

df_modified = df_modified.drop(columns=column_name)

# Display result
print(df_modified)
```

Why did we use columns[3]?

Because:

df.drop(columns="column_name")

normally requires the column name, not its numerical position.

So:

df_modified.columns[3]

finds the name of the column at position 3.

Important viva distinction

drop(index=4) → removes a row

drop(columns="Age") → removes a column

```python

# Q6. CREATE EMPLOYEE DATASET


import pandas as pd

# Employee dataset given in the assignment

data = {
    "Employee_ID": [101, 102, 103, 104, 105],

    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],

    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],

    "Age": [29, 34, 41, 28, 38],

    "Salary": [50000, 70000, 65000, 55000, 60000],

    "Years_of_Experience": [4, 8, 10, 3, 12],

    "Joining_Date": [
        "2020-03-15",
        "2017-07-19",
        "2013-06-01",
        "2021-02-10",
        "2010-11-25"
    ],

    "Gender": ["Female", "Male", "Male", "Female", "Male"],

    "Bonus": [5000, 7000, 6000, 4500, 5000],

    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

# Create DataFrame
employees = pd.DataFrame(data)

# Display DataFrame
print(employees)

# Save as CSV
employees.to_csv("employees.csv", index=False)

print("\nemployees.csv created successfully!")
```

```python

# Q6(f). SORT BY SALARY - DESCENDING


sorted_employees = employees.sort_values(
    by="Salary",
    ascending=False
)

print(sorted_employees)
```

```python

# Q6(g). PERFORMANCE CATEGORY


def performance_category(rating):

    if rating >= 4.5:
        return "Excellent"

    elif rating >= 4.0:
        return "Good"

    else:
        return "Average"


# Apply function to Rating column

employees["Performance_Category"] = employees["Rating"].apply(
    performance_category
)

print(employees)
```

Expected categories
Employee	Rating	Category
Alice	4.5	Excellent
Bob	4.0	Good
Charlie	3.8	Average
Diana	4.7	Excellent
Edward	3.5	Average
Important viva concept
apply()

applies a function to each value of a Series/column.

```python

# Q6(h). FIND MISSING VALUES


print(employees.isnull())

print("\nTotal Missing Values in Each Column:")
print(employees.isnull().sum())
```

What does this mean?
